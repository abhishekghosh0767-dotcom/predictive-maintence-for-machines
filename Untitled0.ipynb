```python
"""
Predictive Maintenance for Machines
Run: python predictive_maintenance.py
Dataset: machine_sensor_data.csv (place in the same folder)
Target: maintenance_required (1 = maintenance/failure risk, 0 = no maintenance flag)
NOTE: This is a synthetic educational dataset, not real industrial telemetry.
"""

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report,
    ConfusionMatrixDisplay, RocCurveDisplay
)

# 1. Load and inspect data
DATA_FILE = "/content/machine_sensor_data-1.csv"
df = pd.read_csv(DATA_FILE)

print("Dataset shape:", df.shape)
print("\nMissing values:\n", df.isnull().sum())
print("\nClass distribution:\n", df["maintenance_required"].value_counts())

# 2. Separate predictors and target
X = df.drop(columns=["maintenance_required"])
y = df["maintenance_required"]

# 3. Split before fitting any preprocessing/model steps.
# Stratification preserves the failure/non-failure ratio in both sets.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 4. Build a pipeline: missing-value handling + classifier.
# class_weight helps when maintenance events are less common.
model = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        max_depth=None,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

# 5. Train
model.fit(X_train, y_train)

# 6. Evaluate on untouched test data
pred = model.predict(X_test)
probability = model.predict_proba(X_test)[:, 1]

print("\n--- TEST SET RESULTS ---")
print(f"Accuracy : {accuracy_score(y_test, pred):.3f}")
print(f"Precision: {precision_score(y_test, pred, zero_division=0):.3f}")
print(f"Recall   : {recall_score(y_test, pred, zero_division=0):.3f}")
print(f"F1-score : {f1_score(y_test, pred, zero_division=0):.3f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, probability):.3f}")
print("\nClassification report:\n", classification_report(y_test, pred, zero_division=0))

# 7. Confusion matrix and ROC curve
ConfusionMatrixDisplay.from_predictions(y_test, pred, cmap="Blues")
plt.title("Predictive Maintenance - Confusion Matrix")
plt.tight_layout()
plt.savefig("confusion_matrix.png", dpi=150)
plt.show()

RocCurveDisplay.from_estimator(model, X_test, y_test)
plt.title("ROC Curve")
plt.tight_layout()
plt.savefig("roc_curve.png", dpi=150)
plt.show()

# 8. Cross-validation on training data only
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_auc = cross_val_score(model, X_train, y_train, cv=cv, scoring="roc_auc", n_jobs=-1)
print(f"\n5-fold CV ROC-AUC: {cv_auc.mean():.3f} +/- {cv_auc.std():.3f}")

# 9. Feature importance (Random Forest)
rf = model.named_steps["classifier"]
importance = pd.Series(rf.feature_importances_, index=X.columns).sort_values()
importance.plot(kind="barh", figsize=(8, 5), title="Feature Importance")
plt.xlabel("Importance")
plt.tight_layout()
plt.savefig("feature_importance.png", dpi=150)
plt.show()

# 10. Example prediction for a new machine reading
new_machine = pd.DataFrame([{
    "temperature_c": 88.0,
    "vibration_mm_s": 4.8,
    "rotational_speed_rpm": 1450.0,
    "pressure_bar": 5.0,
    "tool_wear_minutes": 190,
    "operating_hours": 9500,
    "load_percent": 85.0,
    "days_since_last_maintenance": 130
}])

risk_probability = model.predict_proba(new_machine)[0, 1]
print(f"\nExample machine maintenance-risk probability: {risk_probability:.1%}")
print("Prediction:", "MAINTENANCE FLAG" if risk_probability >= 0.5 else "NO FLAG")

# Important: choose the operational alert threshold using validation data,
# maintenance capacity, and the relative cost of missed failures vs inspections.
```
